```python
import pandas as pd
import plotly.express as px

# 1. Load the dataset saved by your pipeline
df = pd.read_csv("Research_Dataset-candidate_station_table.csv")

# 2. Filter using pre-calculated quality flags
filtered_df = df[df['passes_task1_filters'] == True].copy()

print(f"Original stations: {len(df)} | Usable stations: {len(filtered_df)}")

filtered_df['Status'] = 'Valid Stations (>=30 yrs, <10% missing)'

# 3. Plot with elevation explicitly included in hover data
fig = px.scatter_geo(
    filtered_df,
    lat="latitude",
    lon="longitude",
    color="Status",
    color_discrete_map={'Valid Stations (>=30 yrs, <10% missing)': 'crimson'},
    hover_name="name",
    hover_data={
        "station_id": True,                  # customdata[0]
        "number_of_usable_years": True,      # customdata[1]
        "elevation": True,                   # customdata[2]
        "percent_missing_summer_TMAX": True, # customdata[3]
        "name": False
    },
    scope="usa",
    height=750,
)

# Constrain view strictly to California and immediate neighbors
fig.update_geos(
    lonaxis_range=[-125.5, -113.8],
    lataxis_range=[32.0, 42.5],
    showland=True,
    landcolor="#f8f9fa",
    subunitcolor="#d6d8db",
    countrycolor="#adb5bd",
    oceancolor="#ffffff",
    showocean=True,
    resolution=50
)

# Custom hover template mapping precisely to the customdata array indices
fig.update_traces(
    hovertemplate=(
        "<b>Station:</b> %{hover_name}<br>"
        "<b>Station ID:</b> %{customdata[0]}<br>"
        "<b>Elevation:</b> %{customdata[2]:.1f} m<br>"
        "<b>Usable Years:</b> %{customdata[1]} yrs<br>"
        "<b>TMAX Missing:</b> %{customdata[3]:.1f}%<br>"
        "<extra></extra>"
    ),
    marker=dict(size=6, opacity=0.85, line=dict(width=0.5, color='white'))
)

# Professional academic layout styling
fig.update_layout(
    geo=dict(bgcolor='rgba(0,0,0,0)', framecolor='black', framewidth=1),
    legend=dict(
        title=dict(text="<b>Station Classification</b>"),
        yanchor="bottom", y=0.03, xanchor="left", x=0.03,
        bgcolor="rgba(255, 255, 255, 0.95)", bordercolor="black", borderwidth=1
    ),
    margin={"r":20, "t":20, "l":20, "b":20},
    paper_bgcolor='white',
    plot_bgcolor='white'
)

# North arrow & scale bar annotations
fig.add_annotation(text="N ▲", xref="paper", yref="paper", x=0.95, y=0.92, showarrow=False, font=dict(size=12, color="black"), bgcolor="white", bordercolor="black", borderwidth=1, borderpad=4)
fig.add_annotation(text="Scale ~ 100 km", xref="paper", yref="paper", x=0.95, y=0.85, showarrow=False, font=dict(size=10, color="black"), bgcolor="white", bordercolor="black", borderwidth=1, borderpad=4)

fig.show()
```
